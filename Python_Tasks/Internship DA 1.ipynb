```python
# Q20
a = [10, 20]
b = [10, 20]
print(a == b)
print(a is b)
# OUTPUT:
# True
# False
# Explanation:
# Two separate lists are created with the same elements.
# == checks whether the values of the lists are equal.
# Therefore, a == b returns True.
# is checks whether both variables refer to the same object.
# Since a and b refer to different list objects, a is b returns
# False.

```

```python
# Q21
a = [10, 20]
b = a
print(a == b)
print(a is b)
# OUTPUT:
# True
# True
# Explanation:
# b = a makes b refer to the same list object as a.
# Both variables refer to the same list in memory.
# == returns True because their values are equal.
# is also returns True because they refer to the same object.

```

```python
# Q22
a = "hello"
b = "".join(["he", "llo"])
print(a == b)
print(a is b)
# OUTPUT:
# True
# False
# Explanation:
# a stores the string "hello".
# join() combines "he" and "llo" to create the same text.
# == compares string values, so a == b returns True.
# is checks object identity, not string equality.
# In the usual execution of this code, a and b are different
# string objects, so a is b returns False.
```

```python
# Q23
a = [1]
b = a[:]
print(a == b)
print(a is b)
# OUTPUT:
# True
# False
# Explanation:
# a[:] creates a shallow copy of the list a.
# Both lists contain the same element, 1.
# Therefore, a == b returns True.
# However, slicing creates a new outer list object.
# So, a and b are different objects, and a is b returns False.
```

```python
# Q24
a = int("500")
b = int("500")
print(a == b)
print(a is b)
# OUTPUT:
# True
# False
# Explanation:
# int("500") converts the string "500" into the integer 500.
# Both a and b have the same numerical value.
# Therefore, a == b returns True.
# In the usual CPython execution, these separately created
# integer objects are distinct, so a is b returns False.
# Note: Integer object identity is an implementation detail;
# use == to compare numerical values.
```

```python
# Q25
a = [10, 20]
b = a
b.append(30)
print(a)
# OUTPUT:
# [10, 20, 30]
# Explanation:
# b = a makes both variables refer to the same list object.
# append() adds the element 30 to the existing list.
# Since a and b refer to the same list, the change is visible
# through both variables.
# Therefore, a contains [10, 20, 30].

```

```python
# Q26
a = [10, 20]
b = a
a = a + [30]
print(b)
# OUTPUT:
# [10, 20]
# Explanation:
# Initially, a and b refer to the same list.
# The + operator creates a new list containing [10, 20, 30].
# The variable a is reassigned to this new list.
# The variable b still refers to the original list.
# Therefore, b remains [10, 20].
```

```python
# Q27
a = [10, 20]
b = a
a += [30]
print(b)
# OUTPUT:
# [10, 20, 30]
# Explanation:
# Initially, a and b refer to the same list.
# For lists, += extends the existing list in place.
# The element 30 is added to the original list.
# Since b refers to the same list, the change is visible
# through b as well.
# Therefore, the output is [10, 20, 30].

```

```python
# Q28
a = [10, 20, 30, 40]
b = a[1:3]
print(b)
# OUTPUT:
# [20, 30]
# Explanation:
# List slicing uses the format list[start:end].
# The starting index is included, but the ending index is excluded.
# a[1:3] selects elements at indices 1 and 2.
# These elements are 20 and 30.
# Therefore, b contains [20, 30].

```

```python
# Q29
a = [10, 20, 30, 40]
b = a[::2]
print(b)
# OUTPUT:
# [10, 30]
# Explanation:
# Slicing uses the format list[start:end:step].
# Here, the step value is 2.
# This selects every second element, starting from index 0.
# The selected elements are 10 and 30.
# Therefore, b contains [10, 30].
```

```python
# Q30
a = [10, 20, 30]
b = a[:]
b[0] = 99
print(a)
print(b)
# OUTPUT:
# [10, 20, 30]
# [99, 20, 30]
# Explanation:
# a[:] creates a shallow copy of the list a.
# Therefore, a and b are separate outer list objects.
# b[0] = 99 changes the first element of b only.
# The original list a remains unchanged.
# Hence, the outputs are [10, 20, 30] and [99, 20, 30].
```

```python
# Q35
import copy
a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
# OUTPUT:
# [[1, 99], [3, 4]]
# Explanation:
# copy.copy() creates a shallow copy of the outer list.
# The inner lists are shared between a and b.
# b[0][1] = 99 changes the second element of the first inner list.
# Since the inner list is shared, the change also affects a.
# Therefore, a becomes [[1, 99], [3, 4]].
```

```python
# Q36
import copy
a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
# OUTPUT:
# [[1, 2], [3, 4]]
# Explanation:
# copy.deepcopy() creates a deep copy of the list.
# It creates separate copies of the inner lists as well.
# Therefore, a and b do not share their inner lists.
# Changing b[0][1] does not affect a.
# So, the original list remains unchanged.
```

```python
# Q37
import copy
a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)
# OUTPUT:
# [[10], [20, 30]]
# Explanation:
# a.copy() creates a shallow copy of the outer list.
# The inner lists are shared between a and b.
# a[1].append(30) adds 30 to the second inner list.
# Since that inner list is shared, b also reflects the change.
# Therefore, b prints [[10], [20, 30]].
```

```python
# Q38
import copy
a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)
# OUTPUT:
# [[10], [20]]
# Explanation:
# copy.deepcopy() creates separate copies of the inner lists.
# Therefore, a and b do not share their inner list objects.
# appending 30 to a[1] changes only the list a.
# The list b remains unchanged.
# So, b prints [[10], [20]].

```

```python
# Q39
import copy
a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)
print(a[0] is b[0])
# OUTPUT:
# False
# True
# Explanation:
# copy.copy() creates a new outer list object.
# Therefore, a and b are different outer list objects,
# so a is b returns False.
# However, a shallow copy shares the inner lists.
# Therefore, a[0] and b[0] refer to the same inner list.
# So, a[0] is b[0] returns True.
```

```python
# Q40
a = [[]] * 3
a[1].append(5)
print(a)
# OUTPUT:
# [[5], [5], [5]]
# Explanation:
# [[]] * 3 creates a list containing three references
# to the same inner empty list.
# a[1].append(5) modifies that shared inner list.
# Since all three positions refer to the same list,
# the element 5 appears in every position.
# Therefore, the output is [[5], [5], [5]].
```

```python
# Q44
def test(*args):
    print(args)

test(1, 2)
# OUTPUT:
# (1, 2)
# Explanation:
# *args allows a function to accept any number of positional
# arguments.
# All positional arguments are collected into a tuple named args.
# Here, the values 1 and 2 are passed to the function.
# Therefore, args contains the tuple (1, 2).

```

```python
# Q45
def test(*args):
    print(type(args))

test("a", "b", "c")
# OUTPUT:
# <class 'tuple'>
# Explanation:
# *args collects all positional arguments into a tuple.
# Here, the arguments are "a", "b" and "c".
# The variable args stores these values as a tuple.
# type() returns the data type of args.
# Therefore, the output is <class 'tuple'>.
```

```python

# Q46
def test(**kwargs):
    print(kwargs)

test(name="Ram", age=25)
# OUTPUT:
# {'name': 'Ram', 'age': 25}
# Explanation:
# **kwargs allows a function to accept any number of
# keyword arguments.
# These arguments are collected into a dictionary.
# Here, name is the key with value "Ram", and age is the
# key with value 25.
# Therefore, print() displays the dictionary.
```

```python

# Q47
def test(**kwargs):
    print(type(kwargs))

test(a=10, b=20)
# OUTPUT:
# <class 'dict'>
# Explanation:
# **kwargs collects keyword arguments into a dictionary.
# Here, a and b are the keys, and 10 and 20 are their values.
# type() function returns the data type of kwargs.
# Since kwargs is a dictionary, the output is <class 'dict'>.
```

```python
# Q48
def test():
    print(10)

x = test()
print(x)
# OUTPUT:
# 10
# None
# Explanation:
# The test() function uses print(10) to display the value 10.
# However, the function does not contain a return statement.
# By default, a Python function returns None when no value
# is explicitly returned.
# Therefore, x stores None.
# The first output is 10, and the second output is None.
```

```python
# Q49
def test():
    return 10

x = test()
print(x)
# OUTPUT:
# 10
# Explanation:
# The return statement sends the value 10 back to the caller.
# The function call test() returns 10, which is stored in x.
# print(x) displays the stored value.
# Therefore, the output is 10.
```

```python
# Q50
def test():
    print("Python")
    return 100

x = test()
print(x)
# OUTPUT:
# Python
# 100
# Explanation:
# First, the test() function executes print("Python").
# This displays the word Python.
# Next, return 100 sends the value 100 back to the caller.
# The returned value is stored in the variable x.
# Finally, print(x) displays 100.
# Therefore, the output contains Python followed by 100.
```
